#Criando um banco de dados local

```python
import sqlalchemy
from sqlalchemy import create_engine, MetaData, Table, inspect
engine = create_engine('sqlite:///:memory:')
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/main/clientes_banco.csv'

dados = pd.read_csv(url)

dados.to_sql('clientes', engine, index=False)

inspector = inspect(engine)

print(inspector.get_table_names())
```

```python
query = 'select * from clientes where Categoria_de_renda = "Empregado"'

empregado = pd.read_sql(query, engine)
```

```python
empregado
```

```python
empregado.to_sql('empregados' , con=engine , index=False)
```

```python
pd.read_sql_table('empregados' , engine)
```

```python
pd.read_sql_table('empregados' , engine , columns=['ID_Cliente' , 'Grau_escolaridade', 'Rendimento_anual'])
```

```python
from sqlalchemy import text
```

```python
query = 'DELETE FROM clientes WHERE ID_Cliente=5008804'
with engine.connect() as conn: #abre conexão
    result = conn.execute(text(query)) #executa a query
    conn.commit()  #confirma a alteração
```

```python
query = """UPDATE clientes
 SET Rendimento_anual = 300000
   WHERE ID_CLIENTE = 6840104"""

with engine.connect() as conn:
  result = conn.execute(text(query))
  conn.commit()
```
